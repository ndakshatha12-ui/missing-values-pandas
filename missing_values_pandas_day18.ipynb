# Missing Values in Pandas
Task 18 - Day 18

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Asha", "Raj", "Meena", "Vikram", "Priya", "Karan", "Nisha", "Arjun"],
    "Age": [23, 25, np.nan, 24, 26, 21, np.nan, 23],
    "Department": ["Sales", "IT", "HR", None, "Sales", "HR", "IT", "Sales"],
    "Salary": [45000, np.nan, 38000, 71000, np.nan, 41000, 68000, 47000],
    "Experience": [2, 4, 1, 6, 3, np.nan, 5, 2],
})
df
```

## Missing-value summary

```python
df.isna()
```

```python
print(df.isna().sum())
```

```python
print(df.notna().sum())
```

```python
print("Total missing cells:", df.isna().sum().sum())
```

```python
print(df.isna().mean() * 100, sep="")
```

## Examples using isna() and notna()

```python
# rows where Age is missing
print(df[df["Age"].isna()])
```

```python
# rows where Salary is present (not missing)
print(df[df["Salary"].notna()])
```

## Dropping missing values

```python
# drop any row containing at least one missing value
print(df.dropna())
```

```python
# drop rows missing only in a specific column
print(df.dropna(subset=["Age"]))
```

```python
# drop any column containing at least one missing value
print(df.dropna(axis=1))
```

## Filling missing values

```python
# simplest approach - fill everything with 0
print(df.fillna(0))
```

```python
# a more careful approach - different strategy per column
df_cleaned = df.copy()
df_cleaned["Age"] = df_cleaned["Age"].fillna(df_cleaned["Age"].mean())
df_cleaned["Salary"] = df_cleaned["Salary"].fillna(df_cleaned["Salary"].median())
df_cleaned["Experience"] = df_cleaned["Experience"].fillna(df_cleaned["Experience"].median())
df_cleaned["Department"] = df_cleaned["Department"].fillna("Unknown")

df_cleaned
```

## The cleaned dataset

```python
print(df_cleaned.isna().sum())
```

## Interview questions

**What are missing values?**
Entries in a dataset where no data was recorded for that cell. Pandas represents them as `NaN` (Not a Number) for numeric and object columns, or `None` for general Python objects. They can show up from incomplete data entry, merged datasets with mismatched records, or sensors/surveys that didn't capture every field.

**What are common ways to handle missing values?**
Dropping rows or columns with `dropna()` when missing data is rare and removing it won't bias the analysis; filling with a constant, mean, or median using `fillna()` when the column should keep every row but needs a reasonable placeholder; or filling categorical gaps with a label like "Unknown" to keep the row without guessing a category. The right choice depends on how much is missing and why.

**When should you avoid simply dropping missing rows?**
When missingness is common rather than rare — dropping every row with any gap can throw away a large share of the dataset, as the example above shows (`dropna()` kept only 1 of 8 rows). It's also risky when missing values aren't random: if a particular group tends to have more gaps (e.g. one department didn't report salary), dropping those rows biases the remaining data toward the groups that did report.
